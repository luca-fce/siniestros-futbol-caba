### a) ¿Cuál es la pregunta?

> **¿Se registra un aumento significativo de siniestros viales en la Ciudad de Buenos Aires durante las 12 horas posteriores a la derrota de los equipos más populares de fútbol (Selección Argentina, River Plate y Boca Juniors)?**

Definiciones para que la pregunta sea respondible con los datos:

- **Equipos:** Selección Argentina, River Plate y Boca Juniors, los de mayor convocatoria del país.
- **Partidos:** todos los de 2021 a 2025 incluidos en las bases (568: 501 de River y Boca y 67 de la Selección, oficiales y amistosos).
- **Derrota:** perder el partido, **incluyendo quedar eliminado por penales**.
- **Ventana:** las 12 horas que siguen al final estimado del partido (inicio + 1 h 55 min, más alargue y penales).
- **Siniestros:** todos los registrados por el GCBA en la Ciudad (con víctimas leves, graves o mortales).
- **Aumento significativo:** más siniestros de lo esperado para ese mismo horario, día de la semana y año, con un nivel de significación del 5 %.

**Hipótesis:**
- H0: tras una derrota, los siniestros en las 12 h siguientes no superan lo esperado ni lo observado tras victorias y empates.
- H1: tras una derrota, los siniestros en las 12 h siguientes son mayores.

### b) ¿Por qué vale la pena responderla?

Los siniestros viales son una de las principales causas de muerte evitable en la Ciudad, y la seguridad vial depende tanto de la infraestructura como del **estado de quienes conducen**. La investigación en economía del comportamiento encontró que las derrotas deportivas inesperadas aumentan conductas impulsivas, como la violencia doméstica (Card y Dahl, 2011). Si la frustración tras una derrota de los equipos más populares también se tradujera en una conducción más riesgosa, sería información útil para la gestión pública: reforzar controles de tránsito, campañas de prevención o la planificación de guardias en las horas posteriores a esos partidos. Y si no hay efecto, el dato también sirve, porque evita destinar recursos a un riesgo que no existe.

### c) ¿Qué respuesta encontraron?

**No.** En las 12 horas posteriores a las 127 derrotas analizadas se registraron en promedio 8,07 siniestros, contra 8,12 esperados para ese horario, día y año: **un ratio de 0,99, es decir, lo normal**. Ninguna prueba estadística detecta un aumento (Wilcoxon p = 0,72; Mann-Whitney p = 0,79; regresión de Poisson: −4,1 % frente a las victorias, IC 95 % −11,8 % a +4,4 %, p = 0,34), y el resultado se mantiene en las siete variantes de robustez probadas (sin superclásicos, solo partidos de local en CABA, ventanas de 3 y 6 h, solo siniestros graves y mortales, solo fines de semana). Con estos datos **no se rechaza H0**.

Dos matices: la Selección perdió solo 5 partidos, muy pocos para evaluarla sola, y los seis partidos con mayor exceso de siniestros del período fueron todos **victorias**, entre ellos la final del Mundial 2022 (19 contra 9,7 esperados, casi el doble), un día de festejos masivos en la calle.
