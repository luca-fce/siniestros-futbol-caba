```python
import os
import pandas as pd

# 1. Definir los equipos y las rutas de las carpetas correspondientes
equipos = {
    "River": "datos/Temporadas River - Boca/River Plate/csv",
    "Boca": "datos/Temporadas River - Boca/Boca Juniors/csv"
}

anios = [2021, 2022, 2023, 2024, 2025]

for equipo, ruta_carpeta in equipos.items():
    print(f"Procesando datos para: {equipo}...")
    lista_dfs = []
    prefijo = equipo.lower().replace(" ", "_")

    for anio in anios:
        nombre_archivo = f"{prefijo}_{anio}.csv"
        ruta_archivo = os.path.join(ruta_carpeta, nombre_archivo)

        if os.path.exists(ruta_archivo):
            # Leer el archivo CSV correspondiente
            df = pd.read_csv(ruta_archivo)
            df["temporada_anio"] = anio
            lista_dfs.append(df)
            print(f"  - Cargado: {nombre_archivo} ({len(df)} filas)")
        else:
            print(f"  - Advertencia: No se encontró el archivo {ruta_archivo}")

    if lista_dfs:
        # 2. Unir las tablas en una sola
        df_unificado = pd.concat(lista_dfs, ignore_index=True)

        # 3. Guardar el archivo unificado
        nombre_salida = f"{prefijo}_2021_2025_unificado.csv"
        ruta_salida = os.path.join(ruta_carpeta, nombre_salida)
        df_unificado.to_csv(ruta_salida, index=False)

        print(f"-> ¡Guardado con éxito! {nombre_salida} con un total de {len(df_unificado)} filas.\n")
    else:
        print(f"-> No se encontraron archivos para unificar de {equipo}.\n")

print("¡Proceso finalizado para ambos equipos!")
```
